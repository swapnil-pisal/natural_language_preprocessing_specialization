```python
# our selected sample, complex enouf to eexmplify each step 
tweet = all_positive_tweets[2277]
print(tweet)
```

```python
# download the stopword from NLTK
nltk.download('stopwords')
```

```python
import re                 # Library for regular expression operation
import string       # for string operation

from nltk.corpus import stopwords     # module for stop wordstthat came with nlp
from nltk.stem import PorterStemmer  # module for stemming
from nltk.tokenize import TweetTokenizer   #module for toknize string
```

```python
# remove hyperlinks twitter marks and styles

print('\033[92m' + tweet)
print('\033[94m')

# remove old style retweet text "RT"
tweet2 = re.sub(r'^RT[\s]+', '', tweet)

tweet2 = re.sub(r'https?://[^\s\n\r]+', '', tweet2)

# remove hashtag
# old removing the hash # sign from the word 

tweet2 = re.sub(r'#', '', tweet2)

print(tweet2)
```

```python
# Tokenize the string 
print()
print('\033[92m' + tweet2)
print('\033[94m')

# instantiate tokenizer class 
tokenizer = TweetTokenizer(preserve_case=False, strip_handles=True, reduce_len=True) 

# tokenize tweets 
tweet_tokens = tokenizer.tokenize(tweet2)

print()
print('Tokenized string:')
print(tweet_tokens)
```

```python
# Remove Stop words and punctuations 

# Import the english stop word list from NLTK
stopwords_english = stopwords.words('english')

print('Stop words\n')
print(stopwords_english)

print('\nPunctuation\n') 
print(string.punctuation)
```

```python
print() 
print('\033[92m')
print(tweet_tokens)
print('\033[94m')

tweet_clean = [] 

for word in tweet_tokens:   # go through every word in your tokens list 
    if(word not in stopwords_english and      # remove stopwords 
      word not in string.punctuation):        # remove punctuation
        tweet_clean.append(word)

print('removed stop words and punctuation: ')
print(tweet_clean)
```

```python
# Stemming 

print() 
print('\033[92m')
print(tweet_clean)
print('\033[94m')

# Instantiate stemming class 
stemmer = PorterStemmer()

# create an empty list to store the stems 
tweet_stem = [] 

for word in tweet_clean: 
    stem_word = stemmer.stem(word)       # Stem word
    tweet_stem.append(stem_word)      # append the list

print('stemmed word: ')
print(tweet_stem)
    
    
```
